### 1.1.4 Correlación entre variables numéricas

Se calculará la matriz de correlación entre las variables numéricas utilizando NumPy.


```python
# Seleccionar variables numéricas
columnas_numericas = df.select_dtypes(include=np.number).columns

# Convertir a matriz NumPy
datos_numericos = df[columnas_numericas].to_numpy()

# Calcular matriz de correlación
matriz_correlacion = np.corrcoef(datos_numericos, rowvar=False)

# Mostrar la matriz
pd.DataFrame(
    matriz_correlacion,
    index=columnas_numericas,
    columns=columnas_numericas
)
```

### 1.1.5 Identificación de valores atípicos

Se identificarán posibles valores atípicos en horas_estudio_semanal y monto_inversion utilizando el método del rango intercuartílico (IQR).


```python
# Variables a analizar
variables_outliers = ["horas_estudio_semanal", "monto_inversion"]

for variable in variables_outliers:
    Q1 = df[variable].quantile(0.25)
    Q3 = df[variable].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers = df[
        (df[variable] < limite_inferior) |
        (df[variable] > limite_superior)
    ]

    print(f"\nVariable: {variable}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Límite inferior: {limite_inferior:.2f}")
    print(f"Límite superior: {limite_superior:.2f}")
    print(f"Cantidad de outliers: {len(outliers)}")
```

# Parte 2: Preparación de Datos y Feature Engineering


## 2.1 Transformación de Variables

Se realizarán las transformaciones necesarias para preparar las variables para los modelos de clasificación y regresión: creación de la tasa de completitud y de la variable estudiante_activo, codificación de variables categóricas y estandarización de las variables numéricas.


#### 2.1.1 Creación de nueva variable tasa_completitud

Se calculará la tasa de completitud como el porcentaje de tareas completadas respecto del total de tareas.

```python
# Crear tasa de completitud
df["tasa_completitud"] = (
    df["tareas_completadas"] / df["tareas_totales"]
) * 100

df[["tareas_completadas", "tareas_totales", "tasa_completitud"]].head()
```

#### 2.1.2 Generación de variable binaria estudiante_activo

Se considerará como estudiante activo a quien tenga más de 50 sesiones totales.

```python
# Crear variable binaria estudiante_activo
df["estudiante_activo"] = np.where(
    df["sesiones_totales"] > 50,
    1,
    0
)
```

#### 2.1.3 Codificación de variables categóricas

Las variables categóricas ordinales se codificarán mediante LabelEncoder, mientras que las variables nominales se transformarán mediante OneHotEncoder.

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

# Codificación de la variable ordinal
label_encoder = LabelEncoder()

df["nivel_educacion"] = label_encoder.fit_transform(
    df["nivel_educacion"]
)

# Variables categóricas nominales
variables_nominales = [
    "genero",
    "pais",
    "dispositivo_principal",
    "horario_estudio",
    "experiencia_previa",
    "plataforma_origen"
]

# Aplicar OneHotEncoder
onehot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

nominal_encoded = onehot_encoder.fit_transform(
    df[variables_nominales]
)

# Crear DataFrame con las variables codificadas
nominal_df = pd.DataFrame(
    nominal_encoded,
    columns=onehot_encoder.get_feature_names_out(variables_nominales),
    index=df.index
)

# Reemplazar las variables nominales originales
df = pd.concat(
    [
        df.drop(columns=variables_nominales),
        nominal_df
    ],
    axis=1
)

df.head()
```

#### 2.1.4 Normalización de variables numéricas

Se estandarizarán las variables numéricas predictoras utilizando StandardScaler, de modo que queden en una escala comparable. Las variables objetivo curso_completado y puntaje_satisfaccion no se normalizarán, ya que corresponden a las variables que serán utilizadas para entrenar y evaluar los modelos.

```python
from sklearn.preprocessing import StandardScaler

# Variables numéricas predictoras
columnas_numericas = [
    "edad",
    "horas_estudio_semanal",
    "sesiones_totales",
    "tareas_completadas",
    "tareas_totales",
    "participacion_foros",
    "promedio_evaluaciones",
    "monto_inversion",
    "tasa_completitud",
    "estudiante_activo"
]

# Aplicar StandardScaler solo a las variables predictoras
scaler = StandardScaler()

df[columnas_numericas] = scaler.fit_transform(
    df[columnas_numericas]
)
```

## 2.2 División y Preparación

Se separarán las variables predictoras de las variables objetivo para los modelos de clasificación y regresión. Posteriormente, los datos se dividirán en un 70% para entrenamiento y un 30% para prueba, utilizando random_state=42. Finalmente, se evaluará el desbalance de la variable curso_completado y, si corresponde, se aplicará SMOTE únicamente sobre el conjunto de entrenamiento.

#### 2.2.1 Separación de features y variables objetivo

Se separarán las variables predictoras de las variables objetivo correspondientes a cada tipo de modelo. Para clasificación se utilizará curso_completado, mientras que para regresión se utilizará puntaje_satisfaccion.

```python
# Eliminar identificadores que no aportan información predictiva
df_modelo = df.drop(columns=["id_estudiante", "nombre"])

# Variable objetivo de clasificación
y_clasificacion = df_modelo["curso_completado"]

# Variable objetivo de regresión
y_regresion = df_modelo["puntaje_satisfaccion"]

# Variables predictoras
X = df_modelo.drop(
    columns=["curso_completado", "puntaje_satisfaccion"]
)

print("Dimensiones de X:", X.shape)
print("Dimensiones de y_clasificacion:", y_clasificacion.shape)
print("Dimensiones de y_regresion:", y_regresion.shape)
```

#### 2.2.2 División en entrenamiento y prueba

Los datos se dividirán en un 70% para entrenamiento y un 30% para prueba. Se utilizará random_state=42 para garantizar que la división sea reproducible.

```python
from sklearn.model_selection import train_test_split

# División para clasificación
X_train, X_test, y_clas_train, y_clas_test = train_test_split(
    X,
    y_clasificacion,
    test_size=0.30,
    random_state=42,
    stratify=y_clasificacion
)

# División para regresión
X_train_reg, X_test_reg, y_reg_train, y_reg_test = train_test_split(
    X,
    y_regresion,
    test_size=0.30,
    random_state=42
)

print("Clasificación:")
print("Entrenamiento:", X_train.shape)
print("Prueba:", X_test.shape)

print("\nRegresión:")
print("Entrenamiento:", X_train_reg.shape)
print("Prueba:", X_test_reg.shape)
```

#### 2.2.3 Aplicación de SMOTE

Se evaluará la distribución de la variable curso_completado en el conjunto de entrenamiento. Si existe un desbalance entre las clases, se aplicará SMOTE únicamente sobre los datos de entrenamiento para evitar fuga de información hacia el conjunto de prueba.

```python
from imblearn.over_sampling import SMOTE

# Revisar distribución de las clases
print("Distribución antes de SMOTE:")
print(y_clas_train.value_counts())

print("\nProporción de clases:")
print(y_clas_train.value_counts(normalize=True))
```

##### Interpretación

Se identificó un desbalance en la variable curso_completado, donde la clase 0 representa el 85,43% de las observaciones y la clase 1 el 14,57%, por lo que se aplicará SMOTE al conjunto de entrenamiento

```python
# Aplicar SMOTE únicamente al conjunto de entrenamiento
smote = SMOTE(random_state=42)

X_train_smote, y_clas_train_smote = smote.fit_resample(
    X_train,
    y_clas_train
)

print("Distribución después de SMOTE:")
print(y_clas_train_smote.value_counts())
```

##### Interpretación

Ss aplicó SMOTE únicamente sobre el conjunto de entrenamiento, equilibrando ambas clases para el entrenamiento de los modelos de clasificación.

# Parte 3: Modelado Predictivo con Machine Learning

## 3.1 Modelos de Clasificación


### 3.1.1 RandomForestClassifier

Se entrenará un modelo RandomForestClassifier, basado en un conjunto de árboles de decisión. Para mejorar su desempeño se utilizará GridSearchCV, probando diferentes valores de n_estimators, max_depth y min_samples_split.

La validación cruzada se realizará con 5 particiones (cv=5). Como criterio de selección se utilizará el F1-Score, ya que la variable curso_completado presenta un desbalance entre sus clases y esta métrica considera conjuntamente Precision y Recall.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Crear modelo base
modelo_rf = RandomForestClassifier(
    random_state=42
)

# Definir los hiperparámetros solicitados
parametros_rf = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5, 10]
}

# Configurar GridSearchCV
grid_rf = GridSearchCV(
    estimator=modelo_rf,
    param_grid=parametros_rf,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

# Entrenar utilizando los datos balanceados con SMOTE
grid_rf.fit(
    X_train_smote,
    y_clas_train_smote
)

# Obtener el mejor modelo
mejor_rf = grid_rf.best_estimator_

print("Mejores hiperparámetros de Random Forest:")
print(grid_rf.best_params_)

print("\nMejor F1-Score de validación cruzada:")
print(round(grid_rf.best_score_, 4))
```

#### Interpretación

GridSearchCV evaluó las combinaciones de hiperparámetros definidas para Random Forest utilizando validación cruzada de 5 particiones. La configuración seleccionada corresponde a aquella que obtuvo el mejor F1-Score promedio durante la validación cruzada.

### 3.1.2 LogisticRegression

Se entrenará un modelo de Regresión Logística para predecir la probabilidad de que un estudiante complete el curso.

Mediante GridSearchCV se optimizarán los hiperparámetros C y penalty, utilizando validación cruzada de 5 particiones (cv=5). Al igual que en Random Forest, se utilizará F1-Score como criterio de selección debido al desbalance presente en la variable objetivo.

```python
from sklearn.linear_model import LogisticRegression

# Crear modelo base
modelo_logistic = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Definir los hiperparámetros solicitados
parametros_logistic = {
    "C": [0.01, 0.1, 1, 10, 100],
    "penalty": ["l2"]
}

# Configurar GridSearchCV
grid_logistic = GridSearchCV(
    estimator=modelo_logistic,
    param_grid=parametros_logistic,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

# Entrenar utilizando los datos balanceados con SMOTE
grid_logistic.fit(
    X_train_smote,
    y_clas_train_smote
)

# Obtener el mejor modelo
mejor_logistic = grid_logistic.best_estimator_

print("Mejores hiperparámetros de Logistic Regression:")
print(grid_logistic.best_params_)

print("\nMejor F1-Score de validación cruzada:")
print(round(grid_logistic.best_score_, 4))
```

#### Interpretación

GridSearchCV permitió seleccionar la combinación de C y penalty que obtuvo el mejor F1-Score promedio mediante validación cruzada de 5 particiones. Esta configuración será utilizada para generar las predicciones sobre el conjunto de prueba.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Predicciones Random Forest
y_pred_rf = mejor_rf.predict(X_test)
y_prob_rf = mejor_rf.predict_proba(X_test)[:, 1]

# Predicciones Logistic Regression
y_pred_logistic = mejor_logistic.predict(X_test)
y_prob_logistic = mejor_logistic.predict_proba(X_test)[:, 1]

# Función para calcular métricas
def evaluar_modelo(nombre, y_real, y_pred, y_prob):
    print(nombre)
    print(f"Accuracy:  {accuracy_score(y_real, y_pred):.4f}")
    print(f"Precision: {precision_score(y_real, y_pred):.4f}")
    print(f"Recall:    {recall_score(y_real, y_pred):.4f}")
    print(f"F1-Score:  {f1_score(y_real, y_pred):.4f}")
    print(f"ROC-AUC:   {roc_auc_score(y_real, y_prob):.4f}")
    print("\nMatriz de confusión:")
    print(confusion_matrix(y_real, y_pred))

# Evaluar ambos modelos
evaluar_modelo(
    "Random Forest",
    y_clas_test,
    y_pred_rf,
    y_prob_rf
)

evaluar_modelo(
    "Logistic Regression",
    y_clas_test,
    y_pred_logistic,
    y_prob_logistic
)
```

### 3.1.3 Evaluación de los modelos

Los modelos optimizados se evaluarán utilizando el conjunto de prueba, que se mantuvo sin aplicar SMOTE. Se calcularán las métricas solicitadas: Accuracy, Precision, Recall, F1-Score y ROC-AUC.

Además, se obtendrá la matriz de confusión para identificar la cantidad de verdaderos positivos, verdaderos negativos, falsos positivos y falsos negativos de cada modelo.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)


# Predicciones de Random Forest
pred_rf = mejor_rf.predict(X_test)
prob_rf = mejor_rf.predict_proba(X_test)[:, 1]


# Predicciones de Logistic Regressio
pred_logistic = mejor_logistic.predict(X_test)
prob_logistic = mejor_logistic.predict_proba(X_test)[:, 1]


# Métricas Random Forest
metricas_rf = {
    "Accuracy": accuracy_score(y_clas_test, pred_rf),
    "Precision": precision_score(y_clas_test, pred_rf),
    "Recall": recall_score(y_clas_test, pred_rf),
    "F1-Score": f1_score(y_clas_test, pred_rf),
    "ROC-AUC": roc_auc_score(y_clas_test, prob_rf)
}


# Métricas Logistic Regression
metricas_logistic = {
    "Accuracy": accuracy_score(y_clas_test, pred_logistic),
    "Precision": precision_score(y_clas_test, pred_logistic),
    "Recall": recall_score(y_clas_test, pred_logistic),
    "F1-Score": f1_score(y_clas_test, pred_logistic),
    "ROC-AUC": roc_auc_score(y_clas_test, prob_logistic)
}

# Mostrar métricas
print("MÉTRICAS - RANDOM FOREST")
for metrica, valor in metricas_rf.items():
    print(f"{metrica}: {valor:.4f}")

print("\nMATRIZ DE CONFUSIÓN")
print(confusion_matrix(y_clas_test, pred_rf))


print("\n\nMÉTRICAS - LOGISTIC REGRESSION")
for metrica, valor in metricas_logistic.items():
    print(f"{metrica}: {valor:.4f}")

print("\nMATRIZ DE CONFUSIÓN")
print(confusion_matrix(y_clas_test, pred_logistic))
```

#### Interpretación

En comparación, Random Forest presenta mejores resultados en todas las métricas evaluadas y comete menos errores de clasificación. Por lo tanto, presenta un mejor desempeño para predecir si un estudiante completará el curso.

### 3.2.3 Evaluación de los modelos

Los modelos optimizados se evaluarán utilizando el conjunto de prueba mediante R², MAE y RMSE.

R² mide la proporción de variabilidad del puntaje de satisfacción explicada por el modelo. MAE representa el error absoluto promedio de las predicciones y RMSE mide la magnitud del error dando mayor peso a errores grandes.

### 4.1.2 Crear Columna label

```python
# Crear la variable label a partir de curso_completado
df_spark = df_spark.withColumn(
    "label",
    df_spark["curso_completado"].cast("double")
)

# Verificar distribución de la variable objetivo
df_spark.groupBy("label").count().orderBy("label").show()
```

### 4.1.3 Aplicar StringIndexer a variables categóricas

```python
# Variables categóricas
columnas_categoricas = [
    "nivel_educacion",
    "genero",
    "pais",
    "dispositivo_principal",
    "horario_estudio",
    "experiencia_previa",
    "plataforma_origen"
]

# Crear las columnas indexadas
columnas_indexadas = []

for columna in columnas_categoricas:
    indexador = StringIndexer(
        inputCol=columna,
        outputCol=columna + "_index",
        handleInvalid="keep"
    )

    df_spark = indexador.fit(df_spark).transform(df_spark)
    columnas_indexadas.append(columna + "_index")

# Mostrar resultado
df_spark.select(
    *columnas_categoricas,
    *columnas_indexadas
).show(5)
```

### 4.1.4 Crear columna features con VectorAssembler

```python
# Variables numéricas utilizadas como predictoras
columnas_numericas = [
    "edad",
    "horas_estudio_semanal",
    "sesiones_totales",
    "tareas_completadas",
    "tareas_totales",
    "participacion_foros",
    "promedio_evaluaciones",
    "monto_inversion"
]

# Combinar variables numéricas e indexadas
columnas_features = columnas_numericas + columnas_indexadas

assembler = VectorAssembler(
    inputCols=columnas_features,
    outputCol="features"
)

df_spark = assembler.transform(df_spark)

# Mostrar las variables predictoras
df_spark.select("features", "label").show(5, truncate=False)
```

### 4.1.5 División train/test 70/30

```python
# Dividir en entrenamiento y prueba
train_spark, test_spark = df_spark.randomSplit(
    [0.70, 0.30],
    seed=42
)

print("Registros de entrenamiento:", train_spark.count())
print("Registros de prueba:", test_spark.count())
```

## 5.1 Intervalo de confianza al 95%
Se calculará el intervalo de confianza al 95% para la media de puntaje_satisfaccion. Para ello se utilizará la distribución t de Student, considerando la media, desviación estándar y tamaño de la muestra.


```python
from scipy import stats
import numpy as np

# Datos de puntaje de satisfacción
satisfaccion = df["puntaje_satisfaccion"].dropna()

# Estadísticos
n = len(satisfaccion)
media = satisfaccion.mean()
desv_std = satisfaccion.std()

# Nivel de confianza
confianza = 0.95
alpha = 1 - confianza

# Error estándar
error_estandar = desv_std / np.sqrt(n)

# Valor crítico t
t_critico = stats.t.ppf(1 - alpha / 2, df=n - 1)

# Intervalo de confianza
limite_inferior = media - t_critico * error_estandar
limite_superior = media + t_critico * error_estandar

print("Media de puntaje_satisfaccion:", round(media, 4))
print("Intervalo de confianza al 95%:")
print(f"({limite_inferior:.4f}, {limite_superior:.4f})")
```

## 5.2 Prueba de hipótesis
Se realizará una prueba t para muestras independientes con el objetivo de determinar si existe una diferencia estadísticamente significativa entre el puntaje promedio de satisfacción de los estudiantes que completaron el curso y aquellos que no lo completaron.

Hipótesis nula H0: El puntaje promedio de satisfacción es igual entre los estudiantes que completan y no completan el curso.

Hipótesis alternativa H1: El puntaje promedio de satisfacción es diferente entre los estudiantes que completan y no completan el curso.

Se utilizará un nivel de significancia α = 0,05.


```python
from scipy.stats import ttest_ind

# Separar los grupos según curso completado
satisfaccion_completan = df.loc[
    df["curso_completado"] == 1,
    "puntaje_satisfaccion"
].dropna()

satisfaccion_no_completan = df.loc[
    df["curso_completado"] == 0,
    "puntaje_satisfaccion"
].dropna()

# Prueba t para muestras independientes
estadistico_t, valor_p = ttest_ind(
    satisfaccion_completan,
    satisfaccion_no_completan,
    equal_var=False
)

print("Media - Estudiantes que completan:", round(satisfaccion_completan.mean(), 4))
print("Media - Estudiantes que no completan:", round(satisfaccion_no_completan.mean(), 4))

print("\nEstadístico t:", round(estadistico_t, 4))
print("Valor-p:", round(valor_p, 4))

# Conclusión
alpha = 0.05

if valor_p < alpha:
    print("\nConclusión: Se rechaza H₀.")
    print("Existe una diferencia estadísticamente significativa entre los puntajes promedio de satisfacción.")
else:
    print("\nConclusión: No se rechaza H0.")
    print("No existe evidencia suficiente para afirmar que los puntajes promedio de satisfacción sean diferentes.")
```
